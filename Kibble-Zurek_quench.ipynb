The Hamiltonian for 2D Ising model is
$$
- J(t) \sum_{<i,j>}X_i X_j - g(t) \sum_i Z_i
$$
with $<i,j>$ indicating nearest-neighbor sites on a square lattice.  $Z_i$ and $X_i$ are Pauli matrices on $i$-th site.

The critical value of the field 
$$
g_c = 3.04438
$$

We start the evolution in the state fully polarized in the +$z$ direction, and consider smooth Kibble-Zurek quench, where the couplings are gradually turned on, and the transverse field is turned off.

The evolution goes from $t=0$ to the final $t=t_a$, where $t_a$ is the annealing time, following
$$
J(t) = t / t_a  *  (1- \exp(-32 * t / t_a))
$$
$$
g(t) = g_c * (1 - J(t))
$$

We use $s = t/t_a$ to indicate the fraction of the quench. The critical point is crossed at $s=1/2$.
In the tests, we are interested in annealing times $t_a = 4$ and $8$, monitoring various expectation values along the quench.

A small exponential correction at the begining of the quench protocol makes the first derivative zero, supressing small spurious excitations and allowing for fully adiabatic initialization of the dynamics.
